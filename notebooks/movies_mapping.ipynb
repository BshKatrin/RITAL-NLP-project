```python
import pandas as pd
import numpy as np

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
df = pd.read_csv("../Dataset/predictions/cls_log_reg.csv")
pred = np.where(df["prob_1"] > 0.5, "P", "N")
pd.DataFrame(pred).to_csv("../Dataset/submissions/submission-movie-3.csv", header=None, index=False)
```

```python
true_movies = pd.read_csv("../Dataset/predictions/movie_test.csv", header=None)
```

```python
print(accuracy_score(true_movies, pred))
print(precision_score(true_movies, pred, pos_label="P"))
print(recall_score(true_movies, pred, pos_label="P"))
print(f1_score(true_movies, pred, pos_label="P"))
```
